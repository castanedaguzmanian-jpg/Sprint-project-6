```python
platform_total_sales = df.groupby('platform')['total_sales'].sum().sort_values(ascending=False)
print("Top 10 plataformas por ventas totales:")
print(platform_total_sales.head(10))

platform_first_year = df.groupby('platform')['year_of_release'].min()
print("\nPrimer año con juegos por plataforma:")

platform_last_year = df.groupby('platform')['year_of_release'].max()
print("\nÚltimo año con juegos por plataforma:")

print(platform_first_year.sort_values(ascending=True).head(31))
print(platform_last_year.sort_values(ascending=False).head(31))

# === GRÁFICO QUE TE PIDIERON ===
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.bar(platform_total_sales.index, platform_total_sales.values)

plt.title('Ventas Totales por Plataforma')
plt.xlabel('Plataforma')
plt.ylabel('Ventas Totales (millones)')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
#ventas por anio
import matplotlib.pyplot as plt
import seaborn as sns
top_5_platforms=['PS2','X360','PS3','Wii','DS']
for platform in top_5_platforms:
    temp = df[df['platform'] == platform]
    yearly_sales = temp.groupby('year_of_release')['total_sales'].sum()
    plt.plot(yearly_sales.index, yearly_sales.values, marker='o', label=platform, linewidth=2)

plt.legend()
plt.xlabel("Año")
plt.ylabel("Ventas totales en millones")
plt.title("Distribución de ventas por año (top plataformas)")
plt.grid(True, alpha=0.3)
plt.show()

```

Excelente, muy buen gráfico, permite apreciar el periodo de vida de cada consola, incluyendo su auge y luego caída.

```python
# Calcular la duración del ciclo de vida
platform_lifespan = platform_last_year - platform_first_year + 1
print("\nDuración del ciclo de vida por plataforma (años):")
print(platform_lifespan.sort_values(ascending=False))
mean_platform_lifespan=platform_lifespan.mean()
print (f'\nEl ciclo de vida promedio de las plataformas tiende a ser de: {mean_platform_lifespan:.1f}años')
print("plataformas lider:",top_5_platforms)
```

```python
#compara las ventas de los mismos juegos en otras plataformas.
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Filtrar juegos que aparezcan en más de 5 plataformas
multi_df = df.groupby('name').filter(lambda x: len(x) > 5)

# 2. Calcular ventas totales por juego para elegir los más vendidos
top_games = (
    multi_df.groupby('name')['total_sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)   # <<< seleccionamos 5 juegos con más ventas
    .index
)

# 3. Filtrar dataset original solo a esos juegos top
df_top_games = multi_df[multi_df['name'].isin(top_games)]

# 4. Ordenar: primero por nombre de juego y luego por ventas
df_top_games = df_top_games.sort_values(['name', 'total_sales'], ascending=[True, False])

print(df_top_games.head(20))  # para revisar

# 5. Gráfico comparativo
plt.figure(figsize=(14, 7))
sns.barplot(
    data=df_top_games,
    x='name',
    y='total_sales',
    hue='platform'
)

plt.title('Comparación de Ventas de Juegos Multiplataforma (Top 5 más vendidos)')
plt.xlabel('Nombre del juego')
plt.ylabel('Ventas (millones)')
plt.xticks(rotation=45, ha='right')

plt.legend(title='Plataforma')
plt.tight_layout()
plt.show()
```

```python
#las 5 plataformas principales por region N. America (NA), Union Europea (UE), Japon (JP)
NA_most_sold_platform = df.groupby('platform')['na_sales'].sum().sort_values(ascending=False)
print('Top 5 plataformas mas vendidas en Norte America:')
print(NA_most_sold_platform.head(5))

EU_most_sold_platform = df.groupby('platform')['eu_sales'].sum().sort_values(ascending=False)
print('\nTop 5 plataformas mas vendidas en la Union Europea:')
print(EU_most_sold_platform.head(5))

JP_most_sold_platform = df.groupby('platform')['jp_sales'].sum().sort_values(ascending=False)
print('\nTop 5 plataformas mas vendidas en Japon:')
print(JP_most_sold_platform.head(5))

# ventas combinadas de las 3 regiones
df['three_regions_sales'] = df['na_sales'] + df['eu_sales'] + df['jp_sales']

# top 5 plataformas por estas ventas combinadas
top_5_platforms_3regions = df.groupby('platform')['three_regions_sales'].sum().sort_values(ascending=False).head(5).index

df_top5 = df[df['platform'].isin(top_5_platforms_3regions)]
df_top5.groupby('platform')['na_sales','eu_sales','jp_sales'].mean().plot(kind='bar')
plt.xlabel('plataforma')
plt.ylabel('Ventas Totales (millones)')
plt.title('Top 5 plataformas mas vendidas por region')
plt.show()

#grafica de Norte America
top_5_platforms_na = df.groupby('platform')['na_sales'].sum().sort_values(ascending=False).head(5)
top_5_platforms_na.plot(kind='bar')
plt.xlabel('plataforma')
plt.ylabel('Ventas Totales (millones)')
plt.title('Top 5 plataformas mas vendidas en Norte America')
plt.show()

#grafica de Europa
top_5_platforms_eu = df.groupby('platform')['eu_sales'].sum().sort_values(ascending=False).head(5)
top_5_platforms_eu.plot(kind='bar')
plt.xlabel('plataforma')
plt.ylabel('Ventas Totales (millones)')
plt.title('Top 5 plataformas mas vendidas en Europa')
plt.show()

#grafica de Japon
top_5_platforms_jp = df.groupby('platform')['jp_sales'].sum().sort_values(ascending=False).head(5)
top_5_platforms_jp.plot(kind='bar')
plt.xlabel('plataforma')
plt.ylabel('Ventas Totales (millones)')
plt.title('Top 5 plataformas mas en Japon')
plt.show()
```

```python
#Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son diferentes.
#hipotesis bilateral/dos colas porque busca dferencias en ambos sentidos

action_data = df[df['genre'] == 'Action']['user_score'].dropna()
sports_data = df[df['genre'] == 'Sports']['user_score'].dropna()
alpha = 0.05
results = st.ttest_ind(action_data, sports_data)
print('Valor - p:',results.pvalue)
if(results.pvalue < alpha):
    print("Rechazamos la hipótesis nula")
    print("Las calificaciones promedio de usuarios para los juegos de accion y deportes no son iguales")
else:
    print("No podemos rechazar la hipótesis nula")
    print("Las calificaciones promedio de usuarios para los juegos de accion y deportes son iguales")
    
comparison_data = pd.DataFrame({
    'user_score': list(action_data) + list(sports_data),
    'genre': ['Action'] * len(action_data) + ['Sports'] * len(sports_data)
})

plt.figure(figsize=(8, 6))
sns.boxplot(data=comparison_data, x='genre', y='user_score')
plt.title('Comparación de Calificaciones: Acción vs Deportes')
plt.show()
```

Cómo formulaste las hipótesis nula y alternativa:

*Primera prueba (Xbox One vs PC):
la hipotesis nula y alternativa las formule conforme a lo siguiente:
Hipótesis nula (H₀): Las calificaciones promedio de los usuarios para las plataformas Xbox One y PC son iguales (μ_Xbox = μ_PC)
Hipótesis alternativa (H₁): Las calificaciones promedio de los usuarios para las plataformas Xbox One y PC son diferentes (μ_Xbox ≠ μ_PC)

*Segunda prueba (Acción vs Deportes):
Hipótesis nula (H₀): Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son iguales (μ_Acción = μ_Deportes)
Hipótesis alternativa (H₁): Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son diferentes (μ_Acción ≠ μ_Deportes)

Qué criterio utilizaste para probar las hipótesis y por qué?

*Criterio estadístico utilizado:
Prueba t de Student para muestras independientes (st.ttest_ind())
Nivel de significancia (α) = 0.05

¿Por qué elegiste este criterio?

*Prueba t para muestras independientes: Porque se está comparando las medias de dos grupos diferentes e independientes (Xbox One vs PC, y Acción vs Deportes)
Prueba bilateral (dos colas): Porque se busca detectar diferencias en cualquier dirección, no solo si un grupo es mayor que el otro
α = 0.05: Es el estándar más común en investigación, que significa que se acepta un 5% de probabilidad de cometer un error tipo I (rechazar H₀ cuando es verdadera)

Criterio de decisión:

Si p-valor < α (0.05) → se rechaza H₀
Si p-valor ≥ α (0.05) → No se rechaza H₀
